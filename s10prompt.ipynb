```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.random.gamma(shape=2.0, scale=2.0, size=1000)
```

```python
#bin data
width=0.1
histmin = np.floor(min(x))
histmax = np.ceil(max(x))

bins=np.arange(histmin,histmax,width)

plt.hist(x,bins=100,alpha=0.5, edgecolor='black')
plt.xlabel('number')
plt.ylabel('frequency of number')

plt.savefig('s10prompt.pdf')
```
