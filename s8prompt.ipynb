```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,1,100)
y1 = np.sin(x)
y2 = np.cos(x)
```

```python
fig, axes = plt.subplots(1,2)

#graph the lines
axes[0].plot(x,y1,color='pink',label='sin(x)')
axes[1].plot(x,y2,color='purple',label='cos(x)')

axes[0].legend(fontsize=11)
axes[1].legend(fontsize=11)

plt.savefig('s8prompt.pdf')
```
